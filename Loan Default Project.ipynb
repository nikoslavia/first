```python
#I will keep these and use simple imputer to replace null values of int64 and float64 with median. 
numeric_null_cols = df.select_dtypes(include=['int64', 'float64']).isnull().sum()
numeric_null_cols = numeric_null_cols[numeric_null_cols > 0].index

from sklearn.impute import SimpleImputer
num_imputer = SimpleImputer(strategy='median')
df[numeric_null_cols] = num_imputer.fit_transform(df[numeric_null_cols])
```

```python
#simple imputer for object columns. Using a different strategy here for imputing (most frequent instead of median)

object_null_cols = df.select_dtypes(include=['object']).isnull().sum()
object_null_cols = object_null_cols[object_null_cols > 0].index

from sklearn.impute import SimpleImputer
obj_imputer = SimpleImputer(strategy='most_frequent')
df[object_null_cols] = obj_imputer.fit_transform(df[object_null_cols])
```

```python
#splitting columns for the preprocessor

X=df.drop('TARGET',axis=1)
y=df['TARGET']

categ_cols=X.select_dtypes(include=['object']).columns
robust_cols=X.select_dtypes(include=['int64','float64']).columns
```

```python
#creating a preprocessor

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, RobustScaler

preprocessor = ColumnTransformer([
    ('categorical', OneHotEncoder(handle_unknown='ignore'), categ_cols),
    ('numeric', RobustScaler(), robust_cols)
])
```

```python
preprocessor
```

```python
#fitting our preprocessor into various models

#Logistic Regression

from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)

from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

pipeline_LR=Pipeline([('preprocessor',preprocessor),('model',LogisticRegression(max_iter=1000))])
pipeline_LR.fit(X_train,y_train)
y_pred_LR=pipeline_LR.predict(X_test)
```

```python
#Decision Tree

from sklearn.tree import DecisionTreeClassifier
pipeline_DT=Pipeline([('preprocessor',preprocessor),('model',DecisionTreeClassifier())])
pipeline_DT.fit(X_train,y_train)
y_pred_DT=pipeline_DT.predict(X_test)
```

```python
#XG Boost

from xgboost import XGBClassifier
pipeline_XG=Pipeline([('preprocessor',preprocessor),('model',XGBClassifier())])
pipeline_XG.fit(X_train,y_train)
y_pred_XG=pipeline_XG.predict(X_test)
```

```python
#Random Forest

from sklearn.ensemble import RandomForestClassifier
pipeline_RF=Pipeline([('preprocessor',preprocessor),('model',RandomForestClassifier())])
pipeline_RF.fit(X_train,y_train)
y_pred_RF=pipeline_RF.predict(X_test)
```

```python
#model evaluation
#Model Comparison Report

from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score

results = pd.DataFrame({ "Model": ["Logistic Regression","Decision Tree","Random Forest","XGBoost"],"Accuracy":[accuracy_score(y_test, y_pred_LR),accuracy_score(y_test, y_pred_DT),accuracy_score(y_test, y_pred_RF),accuracy_score(y_test,y_pred_XG)], "Precision": [precision_score(y_test, y_pred_LR, average="weighted"),
precision_score(y_test, y_pred_DT, average="weighted"),precision_score(y_test, y_pred_RF, average="weighted"),precision_score(y_test, y_pred_XG, average="weighted")],"Recall": [recall_score(y_test, y_pred_LR, average="weighted"),recall_score(y_test, y_pred_DT, average="weighted"),recall_score(y_test, y_pred_RF, average="weighted"),
recall_score(y_test, y_pred_XG, average="weighted")],"F1 Score": [f1_score(y_test, y_pred_LR, average="weighted"),f1_score(y_test, y_pred_DT, average="weighted"),
f1_score(y_test, y_pred_RF, average="weighted"),f1_score(y_test, y_pred_XG, average="weighted")]})

results
```

```python
#since accuracy is irrelevant for imbalanced data, classification report is more suitable
from sklearn.metrics import classification_report

print("Logistic Regression")
print(classification_report(y_test, y_pred_LR))

print("Random Forest")
print(classification_report(y_test, y_pred_RF))

print("XGBoost")
print(classification_report(y_test, y_pred_XG))
```

Data Exploration

* The dataset was highly imbalanced, with significantly more non-default cases than default cases.
* There were several numerical and categorical features, with some extreme outliers.
* SK_ID_CURR was dropped as it is only an identifier and has no predictive value.
* Missing values were present in many columns.

Preprocessing

* Missing numerical values were handled using SimpleImputer with the median strategy while I used most frequent strategy for missing categorical values. 
* Categorical features were encoded via OHE before model training.
* RobustScaler was used for numerical features because the dataset contained many outliers.
* The target variable was already in binary format (0 and 1), so no encoding was required.
* The data was split into training and testing sets.

Model Building

* Logistic Regression was used as a baseline model.
* Decision Tree and XGBoost were also used to capture non-linear relationships.
* SVM was considered, but its training time was high due to the large dataset so I abandoned it. 

Model Evaluation

* Accuracy alone was not considered sufficient because of the highly imbalanced target.
* Precision, recall and F1-score were used to evaluate the models, with greater focus on the default class.
* We received poor scores because of imbalanced data but I balanced it later to improve scores.
* XGBOOST was the best performing model despite it's lower accuracy score. It had everything else going for it.
